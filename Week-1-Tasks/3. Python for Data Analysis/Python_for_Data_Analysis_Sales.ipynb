## 7. Load the CSV using Pandas

```python
# Load the supplied CSV file
raw_df = pd.read_csv(dataset_path)

print("Original dataset shape:", raw_df.shape)
display(raw_df.head())
display(raw_df.info())
```

## 8. Inspect the dataset

Use `head`, `tail`, `shape`, `columns`, `dtypes`, `describe`, and missing-value counts to understand the data before cleaning it.

```python
print("Rows and columns:", raw_df.shape)
print("\\nColumn names:")
print(raw_df.columns.tolist())

print("\\nData types:")
print(raw_df.dtypes)

print("\\nMissing values by column:")
display(raw_df.isna().sum().to_frame("Missing_Count"))

print("\\nSummary statistics:")
display(raw_df.describe(include="all").T)
```

## 9. Clean the dataset



```python
df = raw_df.copy()

# 1. Remove rows where every field is empty
before_blank_removal = len(df)
df = df.dropna(how="all")
print("Fully blank rows removed:", before_blank_removal - len(df))

# 2. Remove exact duplicate rows
before_duplicates = len(df)
df = df.drop_duplicates()
print("Duplicate rows removed:", before_duplicates - len(df))

# 3. Clean whitespace in text columns
text_columns = df.select_dtypes(include="object").columns
for column in text_columns:
    df[column] = df[column].str.strip()

# 4. Convert date column
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")

# 5. Convert numeric columns
numeric_columns = ["Quantity", "Unit Price", "Discount", "Sales", "Profit"]
for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

print("Cleaned dataset shape:", df.shape)
display(df.head())
```

## 10. Validate the cleaned dataset

```python
print("Missing values after cleaning:")
display(df.isna().sum().to_frame("Missing_Count"))

print("Duplicate rows remaining:", df.duplicated().sum())
print("Date range:", df["Order Date"].min(), "to", df["Order Date"].max())
print("Unique orders:", df["Order ID"].nunique())
print("Unique customers:", df["Customer ID"].nunique())
print("Unique products:", df["Product"].nunique())

# Optional: inspect any rows that still have missing values
rows_with_missing_values = df[df.isna().any(axis=1)]
display(rows_with_missing_values.head())
```

## 11. NumPy practice



```python
sales_values = df["Sales"].dropna().to_numpy()
profit_values = df["Profit"].dropna().to_numpy()

print("Average sales:", np.mean(sales_values))
print("Median sales:", np.median(sales_values))
print("Sales standard deviation:", np.std(sales_values))
print("Total sales using NumPy:", np.sum(sales_values))

# Calculate row-level profit margin safely (avoid division by zero)
df["Profit Margin (%)"] = np.divide(
    df["Profit"] * 100,
    df["Sales"],
    out=np.zeros(len(df), dtype=float),
    where=df["Sales"].to_numpy() != 0
)

display(df[["Sales", "Profit", "Profit Margin (%)"]].head())
```

## 12. Calculate business KPIs using Pandas

KPIs:
- Total records
- Total unique orders
- Total unique customers
- Total revenue
- Total profit
- Average sales per record
- Overall profit margin
- Total quantity sold

```python
total_records = len(df)
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
total_products = df["Product"].nunique()
total_revenue = df["Sales"].sum()
total_profit = df["Profit"].sum()
average_sales_per_record = df["Sales"].mean()
total_quantity_sold = df["Quantity"].sum()
overall_profit_margin = (total_profit / total_revenue * 100) if total_revenue else 0

kpis = pd.DataFrame({
    "KPI": [
        "Total Records",
        "Unique Orders",
        "Unique Customers",
        "Unique Products",
        "Total Quantity Sold",
        "Total Revenue",
        "Total Profit",
        "Average Sales per Record",
        "Overall Profit Margin (%)"
    ],
    "Value": [
        total_records,
        total_orders,
        total_customers,
        total_products,
        total_quantity_sold,
        total_revenue,
        total_profit,
        average_sales_per_record,
        overall_profit_margin
    ]
})

display(kpis)
```

## 13. Analyse sales by region, category, and product

```python
region_summary = (
    df.groupby("Region", dropna=False)
      .agg(Total_Orders=("Order ID", "nunique"),
           Revenue=("Sales", "sum"),
           Profit=("Profit", "sum"))
      .sort_values("Revenue", ascending=False)
      .round(2)
)

category_summary = (
    df.groupby("Category", dropna=False)
      .agg(Revenue=("Sales", "sum"),
           Profit=("Profit", "sum"),
           Quantity_Sold=("Quantity", "sum"))
      .sort_values("Revenue", ascending=False)
      .round(2)
)

product_summary = (
    df.groupby("Product", dropna=False)
      .agg(Revenue=("Sales", "sum"),
           Profit=("Profit", "sum"),
           Quantity_Sold=("Quantity", "sum"))
      .sort_values("Revenue", ascending=False)
      .round(2)
)

print("Regional summary")
display(region_summary)
print("Category summary")
display(category_summary)
print("Product summary")
display(product_summary)
```

## 14. Monthly sales trend



```python
monthly_summary = (
    df.dropna(subset=["Order Date"])
      .assign(Month=lambda data: data["Order Date"].dt.to_period("M").astype(str))
      .groupby("Month")
      .agg(Revenue=("Sales", "sum"), Profit=("Profit", "sum"))
      .reset_index()
)

display(monthly_summary.round(2))
```

## 15. Visualizations using Matplotlib

The charts below show:
1. Revenue by region.
2. Revenue by product category.
3. Monthly revenue trend.
4. Revenue and profit comparison.

```python
# Chart 1: Revenue by region
plt.figure(figsize=(9, 5))
region_summary["Revenue"].sort_values().plot(kind="barh")
plt.title("Revenue by Region")
plt.xlabel("Revenue")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Revenue by category
plt.figure(figsize=(7, 7))
category_summary["Revenue"].plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Revenue Share by Category")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
# Chart 3: Monthly revenue trend
plt.figure(figsize=(10, 5))
plt.plot(monthly_summary["Month"], monthly_summary["Revenue"], marker="o")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Chart 4: Revenue and profit by region
plot_data = region_summary[["Revenue", "Profit"]]
plot_data.plot(kind="bar", figsize=(10, 5))
plt.title("Revenue and Profit by Region")
plt.xlabel("Region")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
